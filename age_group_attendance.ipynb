```python
def get_age_group(age):
    if age < 0 or age > 110:
        return "Invalid"
    elif age <= 18:
        return "Child"
    elif age < 70:
        return "Adult"
    else:
        return "Old"
```

```python
old_list = []
children_list = []
adults_list = []
```

```python
def get_list(name, number, age):
    checker = get_age_group(age)

    if checker == "Invalid":
        print("The age is invalid.")
        return old_list, children_list, adults_list

    user = {
        "name": name,
        "number": number,
        "age": age
    }

    if checker == "Adult":
        adults_list.append(user)

    elif checker == "Child":
        children_list.append(user)

    elif checker == "Old":
        old_list.append(user)

    return old_list, children_list, adults_list
```

```python
get_attendance = True

while get_attendance:
    start = input("Enter y to start or n to stop: ")

    if start == "y" or start == "Y":
        name, number, age = input(
            "Enter name, number and age e.g [John Doe, 1021021021, 23]: "
        ).split(",")

        name = name.strip()
        number = number.strip()

        try:
            age = int(age.strip())
            get_list(name, number, age)

        except ValueError:
            print("Please enter a valid number for age.")

    elif start == "n" or start == "N":
        get_attendance = False

    else:
        print("Please enter y or n.")
```

```python
print("Old people:")
print(old_list)

print("\nChildren:")
print(children_list)

print("\nAdults:")
print(adults_list)
```

```python
print("\n===== ATTENDANCE SUMMARY =====")

print("\nOld People:")
for person in old_list:
    print(person)

print("\nChildren:")
for person in children_list:
    print(person)

print("\nAdults:")
for person in adults_list:
    print(person)

print("\n===== TOTALS =====")
print("Old people:", len(old_list))
print("Children:", len(children_list))
print("Adults:", len(adults_list))
```
